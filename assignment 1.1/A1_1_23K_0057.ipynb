```python
# 1. ONE-HOT ENCODING

import numpy as np

sentence = "machine learning is useful"

words = sentence.lower().split()

vocabulary = sorted(set(words))

print("Vocabulary:")
print(vocabulary)

one_hot_vectors = np.eye(len(vocabulary))

print("\nOne-Hot Encodings:\n")

for word, vector in zip(vocabulary, one_hot_vectors):
    print(f"{word}: {vector}")
```

```python
# 2. BAG OF WORDS

from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "machine learning is useful",
    "machine learning is powerful",
    "artificial intelligence is useful"
]

vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(documents)

vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

```python
import pandas as pd

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vocabulary
)

print(bow_df)
```

```python
# 3. TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "machine learning is useful",
    "machine learning is powerful",
    "artificial intelligence is useful"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
import pandas as pd

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vocabulary
)

print(tfidf_df.round(3))
```

```python
# 4. WORD2VEC

from gensim.models import Word2Vec

sentences = [
    ["machine", "learning", "is", "useful"],
    ["machine", "learning", "is", "powerful"],
    ["artificial", "intelligence", "is", "useful"],
    ["deep", "learning", "is", "powerful"],
    ["machine", "learning", "uses", "algorithms"],
    ["deep", "learning", "uses", "neural", "networks"]
]

model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1,          # 1 = Skip-gram
    workers=1,
    epochs=100
)

word = "learning"

print(f"Vector for '{word}':")
print(model.wv[word])

print("\nVector size:")
print(len(model.wv[word]))
```

```python
similarity = model.wv.similarity("machine", "learning")

print("Similarity between 'machine' and 'learning':")
print(similarity)
```
